## Overview:
<p>The Hospital Patient EDA Analysis is a study aimed at understanding patient visits, hospital departments, treatments, medical expenses, and patient outcomes. This analysis focuses on finding useful patterns in hospital data and generating insights that can support better decision-making.</p>

## Dataset Description:
* Patient_ID: Unique identifier of the patient.
* Visit_Date: Date of the hospital visit.
* Age: Age of the patient.
* Gender: Gender of the patient.
* Department: Hospital department visited.
* Doctor: Doctor who handled the patient.
* Admission_Type: Type of admission.
* Diagnosis: Medical diagnosis.
* Treatment: Treatment provided to the patient.
* Length_of_Stay: Number of days the patient stayed in the hospital.
* Medical_Expense: Medical expense for the visit.
* Insurance_Type: Type of insurance/payment coverage.
* Heart_Rate: Patient's heart rate.
* Patient_Rating: Rating given by the patient.
* Readmission: Whether the patient was readmitted or not.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```
data = pd.read_csv('Hospital_Patient_EDA_Project.csv.xls')
```

```
data.head()
```

```
data.isnull().sum()
```

```
# Filling missing values
data['Visit_Date'] = data['Visit_Date'].fillna(data['Visit_Date'].mode()[0])

for col in ['Gender', 'Department', 'Doctor', 'Admission_Type',
            'Diagnosis', 'Treatment', 'Insurance_Type']:
    data[col] = data[col].fillna(data[col].mode()[0])

for col in ['Age', 'Length_of_Stay', 'Medical_Expense', 'Heart_Rate', 'Patient_Rating']:
    data[col] = data[col].fillna(data[col].median())
```

```
data.isnull().sum()
```

```
data.info()
```

```
data.duplicated().sum()
```

```
data = data.drop_duplicates()
data.duplicated().sum()
```

```
data['Visit_Date'] = pd.to_datetime(data['Visit_Date'])
data['day_name'] = data['Visit_Date'].dt.day_name()
data['Month'] = data['Visit_Date'].dt.month_name()
```

##### The visit date was converted into a proper date format so that daily and monthly patterns can be studied.

```
admission_count = data['Admission_Type'].value_counts()
admission_count
```

```
admission_count.plot(kind='pie',
                      autopct='%1.1f%%',
                      colors=['#1976D2','#64B5F6','#BBDEFB'],
                      startangle=90,
                      label='')
plt.title('Admission Type of Patients')
plt.show()
```

##### Emergency admissions are slightly higher than regular admissions, while referral admissions are the lowest.

```
department_count = data['Department'].value_counts()
department_count.sort_values().plot(kind='barh')
plt.title('Patients by Department')
plt.xlabel('Number of Patients')
plt.show()
```

##### Neurology has the highest number of patient visits, followed by Orthopedics and Cardiology. General Medicine has the lowest number of visits in this dataset.

```
treatment_count = data['Treatment'].value_counts()
treatment_count.plot(kind='bar')
plt.title('Popular Treatments')
plt.xlabel('Treatment')
plt.ylabel('Number of Patients')
plt.xticks(rotation=45)
plt.show()
```

##### Surgery is the most common treatment, while Procedure is the least common treatment among the recorded visits.

```
diagnosis_count = data['Diagnosis'].value_counts()
diagnosis_count.sort_values().plot(kind='barh')
plt.title('Common Diagnoses')
plt.xlabel('Number of Patients')
plt.show()
```

##### Diabetes is the most frequently recorded diagnosis, followed by Hypertension and Fracture.

```
insurance_count = data['Insurance_Type'].value_counts()
insurance_count.plot(kind='pie',
                     autopct='%1.1f%%',
                     colors=['#1976D2','#64B5F6','#90CAF9','#BBDEFB'],
                     startangle=90,
                     label='')
plt.title('Insurance Type of Patients')
plt.show()
```

##### Self-Pay is the most common insurance type, while Government insurance has the lowest number of patients.

```
department_expense = data.groupby('Department')['Medical_Expense'].sum()
department_expense.sort_values(ascending=False).plot(kind='bar')
plt.title('Total Medical Expense by Department')
plt.xlabel('Department')
plt.ylabel('Medical Expense')
plt.xticks(rotation=45)
plt.show()
```

##### Neurology generates the highest total medical expense among the departments in this dataset.

```
average_stay = data.groupby('Department')['Length_of_Stay'].mean()
average_stay.sort_values(ascending=False).plot(kind='bar')
plt.title('Average Length of Stay by Department')
plt.xlabel('Department')
plt.ylabel('Average Stay (Days)')
plt.xticks(rotation=45)
plt.show()
```

##### Dermatology has the highest average length of stay, while several other departments have average stays close to five days.

```
readmission_count = data['Readmission'].value_counts()
readmission_count.plot(kind='pie',
                       autopct='%1.1f%%',
                       colors=['#1976D2','#90CAF9'],
                       startangle=90,
                       label='')
plt.title('Patient Readmission')
plt.show()
```

##### Most patients were not readmitted. Only a smaller portion of the recorded patients had a readmission.

```
monthly_expense = data.groupby('Month')['Medical_Expense'].sum()
month_order = ['January','February','March','April','May','June',
               'July','August','September','October','November','December']
monthly_expense = monthly_expense.reindex(month_order)

monthly_expense.plot(kind='line', xlabel='', marker='o')
plt.title('Total Medical Expense Over the Months')
plt.xticks(rotation=90)
plt.show()
```

##### March has the highest total medical expense in the available hospital records.

## Conclusion:

- Emergency admission is the most common admission type.
- Neurology has the highest number of patient visits and the highest total medical expense.
- Surgery is the most common treatment in the dataset.
- Diabetes is the most common diagnosis.
- Self-Pay is the most common insurance type.
- Dermatology has the highest average length of stay.
- Most patients were not readmitted.
- March has the highest total medical expense among the recorded months.

## Suggestions:

- Hospitals can plan additional resources for departments with high patient volume, especially Neurology.
- Since emergency admissions are common, emergency departments can maintain sufficient staff and resources during busy periods.
- Popular treatments such as Surgery can be monitored to ensure adequate equipment and staff availability.
- The hospital can study the reasons behind common diagnoses such as Diabetes and Hypertension to improve preventive care.
- Departments with longer average stays can review patient-care processes to identify ways to improve recovery and bed availability.
- Readmission cases can be reviewed to understand possible causes and improve follow-up care.
